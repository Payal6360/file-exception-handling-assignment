1. What is the difference between interpreted and compiled languages?
--> Compiled and interpreted languages differ in how the code is translated and executed. Compiled languages use a compiler to translate the entire source code into machine code before execution. This results in faster performance, as the program runs directly on the system without needing further translation. However, any change in the code requires recompilation. Examples of compiled languages include C, C++, and Go.

In contrast, interpreted languages use an interpreter that reads and executes the code line-by-line at runtime. This makes them more flexible and easier to debug, as errors are identified and displayed immediately during execution. However, interpreted languages tend to be slower in performance because translation happens during runtime. Popular examples include Python, JavaScript, and Ruby.

In summary, compiled languages are typically faster and more efficient but less flexible during development, while interpreted languages are easier to test and modify but may run slower.

2. What is exception handling in Python?
--> Exception handling in Python is a way to manage errors gracefully during program execution. Instead of crashing when an error occurs, Python allows you to catch and handle exceptions using the try, except, else, and finally blocks.

Example:


```
try:
    x = int(input("Enter a number: "))
    print(10 / x)
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Please enter a valid number.")

```

6. What is the difference between import and from ... import in Python?

--> In Python, both import and from ... import are used to access external modules, but they differ in how they import and use those modules. When you use import module, the entire module is loaded, and you must prefix functions or classes with the module name (e.g., math.sqrt(16)). On the other hand, from module import name allows you to import specific functions, classes, or variables directly, so you can use them without the module name (e.g., sqrt(16)). This makes the code shorter, but using import module is generally safer to avoid naming conflicts.

10. What are the advantages of using logging in a program

-->>The advantages of using logging in a program are:
a. Easier Debugging
Logging helps you trace the flow of execution and understand what your program was doing at the time of an error.

Unlike print statements, logs can be filtered by severity (info, warning, error, etc.).

b. Persistent Records
Logs are typically saved to files, so you can look back at them even after the program has stopped running.

c. Better than Print Statements
Logging can include timestamps, log levels, module names, etc.

You can configure where the logs go: console, file, or even remote servers.

d. Helps in Monitoring
In production, logs help monitor the system’s behavior and performance.

Useful for spotting patterns, bottlenecks, or anomalies.

e. Customization & Control
You can adjust the log level to control how much detail you want (e.g., DEBUG, INFO, WARNING, ERROR, CRITICAL).

Logging can be turned on or off for different parts of the application.

f. Thread & Multiprocessing Safe
The logging module in Python (and similar in other languages) is safe to use across threads and processes, unlike manual print statements.

18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?


````python
from logging.handlers import RotatingFileHandler
rotating_logger = logging.getLogger("RotatingLogger")
handler = RotatingFileHandler("rotated.log", maxBytes=1024*1024, backupCount=3)
rotating_logger.addHandler(handler)
rotating_logger.setLevel(logging.INFO)
rotating_logger.info("This is a log message with rotation.")
````

21. Write a Python program that reads a file and prints the number of occurrences of a specific word.


````python
word = "test"
count = 0
with open("example.txt", "r") as file:
    for line in file:
        count += line.lower().count(word.lower())
print(f"The word '{word}' occurred {count} times.")
````
